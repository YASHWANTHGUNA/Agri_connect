```python
import pandas as pd
```

```python
housingdf = pd.read_csv('/content/housing.csv')
housingdf
```

```python
X = housingdf['Area Population'].tolist()
Y = housingdf['Price'].tolist()
```

Simple Linear Regression without Library.

```python
mean_x = sum(X)/len(X)
mean_y = sum(Y)/len(Y)
```

```python
sum_num = 0
sum_denom = 0
for i in range(len(X)):
  num=(X[i] -mean_x) * (Y[i]-mean_y)
  sum_num += num
  denom = (X[i] - mean_x) ** 2
  sum_denom += denom
```

```python
m=sum_num/sum_denom
print(m)
c=mean_y - m*mean_x
print(c)
```

```python
y_pred = [m*x + c for x in X]
y_pred
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,Y)
plt.plot(X,y_pred, color='red')

plt.show()
```

Using Library

```python
X=housingdf['Area Population']
Y=housingdf['Price']
print(type(X))
print(type(Y))
```

import numpy


```python
import numpy as np
```

```python
X = np.array(X).reshape(-1,1)
Y = np.array(Y)
print(type(X))
print(type(Y))
print(X)
print('BREAK')
print(Y)
```

import library

```python
from sklearn.linear_model import LinearRegression
```

Making Model

```python
model = LinearRegression()
model.fit(X,Y)
m = model.coef_[0]
print(m)
c = model.intercept_
print(c)
```

```python
y_pred = [m*x + c for x in X]
y_pred
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, Y, label = 'Actual Data')
plt.plot(X, y_pred, color='red', label = 'Best Fit Line')
plt.show()
```

Multiple Linear Regression

Dropping a Column

```python
housingdf_copy = housingdf.copy()
housingdf
```

```python
housingdf_copy.drop(columns={'Price','Address'}, axis=1,inplace=True)
housingdf_copy
```

```python
X=np.array(housingdf_copy)
X
```

```python
Y=np.array(housingdf['Price'])
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.25,random_state=23)
```

```python
model=LinearRegression()
model.fit(X_train,Y_train)
y_pred=model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import r2_score,mean_squared_error
r2=r2_score(Y_test,y_pred)
r2
```

```python
mse=mean_squared_error(Y_test,y_pred)
mse
```

## Visualization of Multiple Linear Regression Results

```python
# Since we have multiple features, we can't visualize the results in a simple 2D scatter plot like with simple linear regression.
# However, we can visualize the predicted vs actual prices.
plt.scatter(Y_test, y_pred)
plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")
plt.title("Actual Prices vs Predicted Prices")
plt.show()
```
